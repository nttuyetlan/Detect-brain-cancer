```python
import os
import shutil

def move_dataset(source_dir, destination_dir):
    """
    Chuyển một dataset từ thư mục nguồn sang thư mục đích.
    
    :param source_dir: Đường dẫn đến thư mục chứa dataset nguồn.
    :param destination_dir: Đường dẫn đến thư mục đích nơi dataset sẽ được chuyển đến.
    """
    # Kiểm tra xem thư mục nguồn có tồn tại không
    if not os.path.exists(source_dir):
        print(f"Thư mục nguồn không tồn tại: {source_dir}")
        return
    
    # Tạo thư mục đích nếu nó không tồn tại
    os.makedirs(destination_dir, exist_ok=True)

    # Lặp qua từng file trong thư mục nguồn và chuyển chúng đến thư mục đích
    for item in os.listdir(source_dir):
        source_path = os.path.join(source_dir, item)  # Đường dẫn đầy đủ đến file
        destination_path = os.path.join(destination_dir, item)  # Đường dẫn đích
        if os.path.isdir(source_path):
            shutil.copytree(source_path, destination_path)  # Sao chép thư mục
        else:
            shutil.copy(source_path, destination_path)  # Sao chép file
        # Nếu bạn muốn xóa file gốc, hãy uncomment dòng sau:
        # os.remove(source_path)

    print(f"Đã chuyển tất cả các file từ {source_dir} đến {destination_dir}")

# Ví dụ sử dụng
move_dataset('/kaggle/input/brain-tumor-by-plane-object-detection/axial_t1wce_2_class/test/labels', '/kaggle/working/brain_tumor_object_detection/axial/val/labels')
move_dataset('/kaggle/input/brain-tumor-by-plane-object-detection/axial_t1wce_2_class/train/labels', '/kaggle/working/brain_tumor_object_detection/axial/val/labels')
move_dataset('/kaggle/input/brain-tumor-by-plane-object-detection/coronal_t1wce_2_class/test/labels', '/kaggle/working/brain_tumor_object_detection/coronal/val/labels')
move_dataset('/kaggle/input/brain-tumor-by-plane-object-detection/coronal_t1wce_2_class/train/labels', '/kaggle/working/brain_tumor_object_detection/coronal/val/labels')
move_dataset('/kaggle/input/brain-tumor-by-plane-object-detection/sagittal_t1wce_2_class/test/labels', '/kaggle/working/brain_tumor_object_detection/sagittal/val/labels')
move_dataset('/kaggle/input/brain-tumor-by-plane-object-detection/sagittal_t1wce_2_class/train/labels', '/kaggle/working/brain_tumor_object_detection/sagittal/val/labels')
```
